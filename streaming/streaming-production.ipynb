```python
path = '/Volumes/spark_definitive_guide/spark_book/activity-data/'
static = spark.read.json(path)
```

```python
streaming = spark\
    .readStream\
    .schema(static.schema)\
    .option("maxFilesPerTrigger", 10)\
    .json(path)\
    .groupBy("gt")\
    .count()
query = streaming\
    .writeStream\
    .outputMode("complete")\
    .trigger(availableNow=True)\
    .option("checkpointLocation", "/Volumes/spark_definitive_guide/spark_book/activity-data/checkpoints/test_python_stream")\
    .queryName("test_python_stream")\
    .format("memory")\
    .start()
```
